**Question 1: What are Large Language Models (LLMs) and how do they function?**

Answer:

* Large Language Models (LLMs) are AI models trained on **huge amounts of text data** to understand and generate human-like language.

* They are based on **Large Language Models**, which learn patterns, grammar, and context from data.

* **How they work:**

  * They break text into tokens (words or subwords).
  * Each token is converted into numerical vectors (embeddings).
  * The model predicts the **next word in a sequence** repeatedly to generate text.
  * They use architectures like **transformers** with attention mechanisms.

* **Output:**

  * Can generate text, answer questions, summarize, translate, and more.

**Question 5: Compare and contrast LangChain and LlamaIndex. What unique problems does each solve?**

Answer:

* **LangChain**

  * Focuses on building **end-to-end LLM applications**
  * Helps in connecting LLMs with tools, APIs, memory, and agents
  * Useful for creating **chatbots, agents, and multi-step workflows**
  * Solves the problem of **orchestrating multiple components together**

* **LlamaIndex**

  * Focuses on **connecting LLMs with external data sources**
  * Helps in indexing, retrieving, and querying private data (PDFs, documents, databases)
  * Solves the problem of **efficient data retrieval for LLMs (RAG systems)**

**Key Difference:**

* LangChain → focuses on **application logic and workflow building**
* LlamaIndex → focuses on **data ingestion and retrieval from external knowledge sources**

**Simple Example:**

* LangChain → building a chatbot with tools and memory
* LlamaIndex → making that chatbot understand your personal PDFs or documents

**Conclusion:**

* Both are often used together:

  * LlamaIndex for **data retrieval (RAG layer)**
  * LangChain for **application orchestration (agent layer)** 🚀

```python
'''
Question 6: Implement a basic Langchain pipeline using OpenAI’s LLM to answer
questions based on a user input prompt.
'''

import os
from langchain_openai import ChatOpenAI
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
import getpass

# STEP 1: Enter API key safely
os.environ["OPENAI_API_KEY"] = getpass.getpass("Enter your OpenAI API key: ")

# STEP 2: Initialize model
llm = ChatOpenAI(model="gpt-3.5-turbo")

# STEP 3: Prompt template
prompt = ChatPromptTemplate.from_template(
    "Answer the following question in a simple and clear way:\n{question}"
)

# STEP 4: Output parser
parser = StrOutputParser()

# STEP 5: Create pipeline
chain = prompt | llm | parser

# STEP 6: User input
question = input("Enter your question: ")

# STEP 7: Get response
response = chain.invoke({"question": question})

# STEP 8: Print answer
print("\nAnswer:\n", response)
```

```python
'''
Question 7: Integrate Langchain with a third-party API (e.g., weather, news) and show
how responses can be generated via LLMs.

'''


import requests
from transformers import pipeline

# Step 1: Fetch weather data (API)
url = "https://api.open-meteo.com/v1/forecast?latitude=19.07&longitude=72.87&current_weather=true"
weather = requests.get(url).json()["current_weather"]

# Step 2: Load a stable model
generator = pipeline("text-generation", model="gpt2")

# Step 3: Create structured prompt
prompt = f"""
You are a weather assistant.
Explain the following weather data in simple human language:

Temperature: {weather['temperature']}°C
Wind Speed: {weather['windspeed']} km/h
Wind Direction: {weather['winddirection']}°
Weather Code: {weather['weathercode']}

Answer:
"""

# Step 4: Generate output
result = generator(prompt, max_new_tokens=80)

# Step 5: Print result
print("\nFINAL ANSWER:\n")
print(result[0]["generated_text"])
```

```python
'''
Question 9: Demonstrate combining Langchain with LamaIndex to create a simple
document-based Q&A chatbot.
'''
# =========================
# STEP 1: Install libraries
# =========================
!pip install llama-index sentence-transformers -q

# =========================
# STEP 2: Create sample file
# =========================
with open("data.txt", "w") as f:
    f.write("""
Artificial Intelligence is the simulation of human intelligence in machines.
Machine Learning is a subset of AI that allows systems to learn from data.
NLP helps machines understand human language.
""")

print("File created successfully!")

# =========================
# STEP 3: Imports
# =========================
from llama_index.core import VectorStoreIndex, SimpleDirectoryReader, Settings
from llama_index.embeddings.huggingface import HuggingFaceEmbedding

# =========================
# STEP 4: Disable LLM (NO API)
# =========================
Settings.llm = None

# =========================
# STEP 5: Load documents
# =========================
documents = SimpleDirectoryReader(".").load_data()

# =========================
# STEP 6: Load local embedding model
# =========================
embed_model = HuggingFaceEmbedding(
    model_name="sentence-transformers/all-MiniLM-L6-v2"
)

# =========================
# STEP 7: Create index
# =========================
index = VectorStoreIndex.from_documents(
    documents,
    embed_model=embed_model
)

# =========================
# STEP 8: Create retriever (IMPROVED)
# =========================
retriever = index.as_retriever(similarity_top_k=3)

# =========================
# STEP 9: Chatbot loop
# =========================
print("\n🤖 Chatbot Ready! Type 'exit' to stop.\n")

while True:
    question = input("You: ")

    if question.lower() == "exit":
        print("Chatbot: Goodbye!")
        break

    nodes = retriever.retrieve(question)

    answer = ""

    # Smart filtering logic
    for node in nodes:
        text = node.text.lower()
        q = question.lower()

        if ("ml" in q or "machine learning" in q) and "machine learning" in text:
            answer = node.text
        elif ("nlp" in q) and "nlp" in text:
            answer = node.text
        elif ("ai" in q or "artificial intelligence" in q) and "artificial intelligence" in text:
            answer = node.text

    # fallback if nothing matched
    if answer == "":
        answer = nodes[0].text

    print("Chatbot:", answer)
```
